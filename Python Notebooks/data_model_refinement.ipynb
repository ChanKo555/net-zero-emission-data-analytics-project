# Additional Data Transformation

In this notebook, we will do the transformations discussed in EDA.

```python
# Import libraries and read in data
import pandas as pd

elec_path = r"C:\Users\Admin\Desktop\net-zero emission by 2025 project\net-zero-emission-data-analytics-project\cleaned datasets\elec_gen_and_consump.csv"
ghg_path = r"C:\Users\Admin\Desktop\net-zero emission by 2025 project\net-zero-emission-data-analytics-project\cleaned datasets\ggas_emissions.csv"

elec_gen_and_consump = pd.read_csv(elec_path)
ggas_emissions = pd.read_csv(ghg_path)
```

```python
elec_gen_and_consump.head()
```

```python
ggas_emissions.head()
```

In both of these datasets, we can see that there is total, and there's another description for each dataset. Therefore, we will split out rows that represent total value to as separate datsets. First, we will begin with electricity generation and consumption dataset.

```python
# Separate total electricity generation and consumption
elec_gen_and_consump_total = elec_gen_and_consump[
    elec_gen_and_consump['category'] == 'Total'
].copy()

# Remove the category column
elec_gen_and_consump_total = elec_gen_and_consump_total.drop(
    columns=['category']
)

# Preview
display(elec_gen_and_consump_total.head())
```

Now we will remove total rows in the original dataset, and also remove the category column since all rows represent Consumption Sector now.

```python
# Separate consumption sector data
elec_gen_and_consump_sector = elec_gen_and_consump[
    elec_gen_and_consump['category'] != 'Total'
].copy()

# Remove the category column
elec_gen_and_consump_sector = elec_gen_and_consump_sector.drop(
    columns=['category']
)

# Preview
display(elec_gen_and_consump_sector.head())
```

To make sure there are no missing or duplicate values, we will perform a check in the following cell.

```python
# Check missing values
print("Missing values - elec_gen_and_consump_total:")
print(elec_gen_and_consump_total.isnull().sum())

print("\nMissing values - elec_gen_and_consump_sector:")
print(elec_gen_and_consump_sector.isnull().sum())


# Check duplicate rows
print("\nDuplicate rows - elec_gen_and_consump_total:")
print(elec_gen_and_consump_total.duplicated().sum())

print("\nDuplicate rows - elec_gen_and_consump_sector:")
print(elec_gen_and_consump_sector.duplicated().sum())
```

Next, we will move on to doing similar transformations to greenhouse gas emissions dataset.

```python
ggas_emissions
```

```python
# Separate total GHG emissions
ggas_emissions_total = ggas_emissions[
    ggas_emissions['category'] == 'Total'
].copy()

# Remove category column from total dataset
ggas_emissions_total = ggas_emissions_total.drop(
    columns=['category']
)

# Remove total rows from original dataset, keeping only gas types
ggas_emissions_gas_type = ggas_emissions[
    ggas_emissions['category'] != 'Total'
].copy()

# Remove category column from gas-type dataset
ggas_emissions_gas_type = ggas_emissions_gas_type.drop(
    columns=['category']
)
```

```python
display(ggas_emissions_total.head())
```

```python
display(ggas_emissions_gas_type.head())
```

```python
# Check missing values

print("Missing values - ggas_emissions_total:")
print(ggas_emissions_total.isnull().sum())

print("\nMissing values - ggas_emissions_gas_type:")
print(ggas_emissions_gas_type.isnull().sum())

# Check duplicate rows

print("\nDuplicate rows - ggas_emissions_total:")
print(ggas_emissions_total.duplicated().sum())

print("\nDuplicate rows - ggas_emissions_gas_type:")
print(ggas_emissions_gas_type.duplicated().sum())
```

The results above show transformation is successfully done and the datasets are clean. Now we will export the cleaned datasets as csv files.

```python
# Export transformed datasets as CSV
output_path = r"C:\Users\Admin\Desktop\net-zero emission by 2025 project\net-zero-emission-data-analytics-project\cleaned datasets"

elec_gen_and_consump_total.to_csv(
    output_path + r"\elec_gen_and_consump_total.csv",
    index=False
)

elec_gen_and_consump_sector.to_csv(
    output_path + r"\elec_gen_and_consump_sector.csv",
    index=False
)

ggas_emissions_total.to_csv(
    output_path + r"\ggas_emissions_total.csv",
    index=False
)

ggas_emissions_gas_type.to_csv(
    output_path + r"\ggas_emissions_gas_type.csv",
    index=False
)
```
